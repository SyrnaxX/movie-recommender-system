## **Word Prepropcessing**

```python
# import nlp libraries
import nltk
from nltk.corpus import stopwords
from nltk.stem  import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords',quiet=True)
nltk.download('wordnet',quiet=True)
```

```python
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()
```

```python
def clean_text(text):
  text = re.sub('[^a-zA-Z]',' ',text) # removing punctuation
  text = text.lower() # converting in lower case
  text = text.split() # tokenization in words
  text = [lemmatizer.lemmatize(word) for word in text if word not in stop_words] # removing stop words and lemmatize words
  text = " ".join(text)
  return text
```

```python
data['tags'] = data['tags'].apply(clean_text)
```

```python
data.iloc[0]['tags']
```

```python
## reset index
data = data.reset_index(drop=True)
```

```python
indices = pd.Series(data.index, index=data['title'])
indices
```

# **Vectorization**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tfidf = TfidfVectorizer(max_features=50000,ngram_range=(1,2),stop_words='english')
tfidf_matrix = tfidf.fit_transform(data['tags'])
tfidf_matrix.shape
```

## **Cosine Similarity**

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
def recommend(title:str,n=10):
  if title not in indices:
    return ['Movie not found!!']
  idx = indices[title]
  similarity_scores = cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  similar_idx = similarity_scores.argsort()[::-1][1:n+1]
  return data.iloc[similar_idx]['title']
```

```python
recommend('Avenger',7)
```

```python
import joblib
joblib.dump(tfidf,'tfidf.pkl')
joblib.dump(indices,'indices.pkl')
joblib.dump(data,'data.pkl')
joblib.dump(tfidf_matrix,'tfidf_matrix.pkl')
```
